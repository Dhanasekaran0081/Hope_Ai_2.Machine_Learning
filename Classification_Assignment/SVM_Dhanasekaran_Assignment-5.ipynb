```python
# Storing value counts in a dictionary where key = column name
object_counts = {col: df[col].value_counts() for col in df.select_dtypes(include='object').columns}

# Example: Access the counts for a specific column named 'sg'
# print(object_counts['sg'])
```

```python
# This converts paired categories cleanly into 1s and 0s
binary_mappings = {
    "normal": 1,
    "abnormal": 0,
    "present": 1,
    "notpresent": 0,
    "yes": 1,
    "no": 0,
    "good": 1,
    "poor": 0,
}
```

```python
# Creating a list of columns that strictly contain these binary classifications
binary_cols = ["rbc", "pc", "pcc", "ba", "htn", "dm", "cad", "appet", "pe", "ane", "classification"]
```

```python
# Applying the dictionary mapping to replace the text with 1 and 0 integers
df[binary_cols] = df[binary_cols].replace(binary_mappings)
```

```python
# Automatically detect any remaining text columns (like 'sg' which has a, b, c, d, e)
# select_dtypes(include='object').columns isolates unhandled text data fields
remaining_objects = df.select_dtypes(include=["object"]).columns.tolist()
```

```python
# Applying One-Hot Encoding to multi-category columns using pd.get_dummies()
# columns=remaining_objects specifies which columns to split up
# dtype=int forces the output cells to map as integers (1 and 0) instead of (True and False)
# drop_first=True prevents multi-collinearity (drops 1 dummy column to avoid mathematical redundancies)
df_final = pd.get_dummies(df, columns=remaining_objects, dtype=int, drop_first=True)
```

```python
# Printing out the final dataset to verify all object columns successfully became integers
print(df_final.head())
print("\n--- Current Data Types After Conversion ---")
print(df_final.dtypes)
```

```python
df_final.columns
```

```python
#Storing independent variables as 'X' and dependent variable as 'y'

X = df_final[['age', 'bp', 'al', 'su', 'rbc', 'pc', 'pcc', 'ba', 'bgr', 'bu', 'sc',
       'sod', 'pot', 'hrmo', 'pcv', 'wc', 'rc', 'htn', 'dm', 'cad', 'appet',
       'pe', 'ane', 'sg_b', 'sg_c', 'sg_d', 'sg_e']]
y = df_final['classification']
```

```python
X.shape
```

```python
#splitting into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 1/3, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.svm import SVC
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'kernel':['linear','rbf','poly','sigmoid'],
             'gamma':['auto','scale'],
             'C':[10,100,1000,2000,3000]} 



grid = GridSearchCV(SVC(probability=True), param_grid, refit = True, verbose = 3,n_jobs=-1,scoring='f1_weighted') 
   
# fitting the model for grid search 
grid.fit(X_train, y_train)
```

```python
# print best parameter after tuning 
#print(grid.best_params_) 
re=grid.cv_results_
#print(re)
grid_predictions = grid.predict(X_test) 
   

from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, grid_predictions)



# print classification report 
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, grid_predictions)
```

```python
from sklearn.metrics import f1_score
f1_macro=f1_score(y_test,grid_predictions,average='weighted')
print("The f1_macro value for best parameter {}:".format(grid.best_params_),f1_macro)
```

```python
print("The confusion Matrix:\n",cm)
```

```python
print("The report:\n",clf_report)
```

```python
from sklearn.metrics import roc_auc_score

roc_auc_score(y_test,grid.predict_proba(X_test)[:,1])
```

```python
#Result: We are getting an 'f1-score_weighted average' of 99%. It is a Excellent score
```

```python
# 1. Get the predicted probabilities for the positive class (Class 1: Diseased)
# Note: Ensure your SVC model was initialized with probability=True
y_probs_svm = grid.predict_proba(X_test)[:, 1]

# 2. Lower the threshold from 0.5 to 0.42 
# (Since AUC is 1.0, this safely eliminates the False Negative without introducing False Positives)
custom_threshold = 0.42
new_predictions = (y_probs_svm >= custom_threshold).astype(int)

# 3. Generate the updated confusion matrix and report
new_cm = confusion_matrix(y_test, new_predictions)
new_report = classification_report(y_test, new_predictions)

print(f"New Confusion Matrix (Threshold = {custom_threshold}):\n", new_cm)
print("\nNew Classification Report:\n", new_report)
```

```python
import joblib

# 1. Capture the exact column blueprint expected by the model/scaler
training_column_layout = X.columns.tolist()

# 2. Package everything neatly
model_payload = {
    'model': grid.best_estimator_,  
    'threshold': 0.42,
    'scaler': sc,
    'binary_mappings': binary_mappings,
    'binary_cols': ['rbc', 'pc', 'pcc', 'ba', 'htn', 'dm', 'cad', 'appet', 'pe', 'ane'], # Note: excludes classification target
    'expected_columns': training_column_layout
}

# 3. Save to a production pipeline file
joblib.dump(model_payload, 'svm_ckd_production_pipeline.pkl')
print("Production pipeline assets successfully exported!")
```
