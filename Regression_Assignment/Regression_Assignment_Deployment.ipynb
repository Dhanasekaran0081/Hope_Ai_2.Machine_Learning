```python
import numpy as np
import pickle
```

```python
model = pickle.load(open('finalized_RF_model.sav', 'rb'))
```

```python
#Loading the saved model and scaler
with open("rf_model.pkl", "rb") as f:
    loaded_model = pickle.load(f)

with open("scaler.pkl", "rb") as f:
    loaded_scaler = pickle.load(f)
```

```python
age_input=int(input("Enter your age:"))
bmi_input=float(input("Enter your bmi value:"))
children_input=int(input("Enter number of childrens:"))
sex_input=int(input("Enter your sex, if male '1' or else '0':"))
smoker_input=int(input("Do you Smoke?, if Yes '1' or else '0':"))

#Shaping the input into a 2D array (1 row, multiple columns)
raw_input = np.array([[age_input, bmi_input, children_input, sex_input, smoker_input]])

#Transforming the raw input using the training scaler
scaled_input = loaded_scaler.transform(raw_input)

result=loaded_model.predict(scaled_input)

print("Your Insurance Charges will be ", result)
```
