#### Step:1

```python
#'Smoker' & 'Sex' features are Catagorical. For Machine Learning purpose, i am converting those features into numerical variables
df=pd.get_dummies(df,drop_first=True,dtype=int)
```

```python
df
```

```python
df.columns
```

#### Step:2

```python
#splitting Input and Output Variables
independent=df[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dependent=df[['charges']]
```

```python
independent
```

```python
dependent
```

#### Step:3

```python
#split into training set and test (70/30 ratio)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(independent, dependent, test_size = 0.3, random_state = 0)
```

### 4.) Develop a good model with r2_score. You can use any machine learning algorithm; you can create many models. Finally, you have to come up with final model.

> **Machine Learning Algorithms:**

#### 1. Multiple Linear Regression

```python
#importing LinearRegression from sklearn library
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
#predicting the X_test with the above model
y_pred=regressor.predict(X_test)
```

```python
#Model Evaluation
from sklearn.metrics import r2_score
r_score=r2_score(y_pred,y_test)
round(r_score,3)
```

r2 score is low, **0.71**.

#### 2. Support Vector Machine

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
#importing LinearRegression from sklearn library
from sklearn.svm import SVR
regressor_svm = SVR(kernel='rbf',C=3000)
regressor_svm.fit(X_train, y_train)
```

```python
y_pred=regressor_svm.predict(X_test)
```

```python
#Model Evaluation
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
round(r_score,3)
```

> The r2 score is better, **0.866**.

#### 3. Decision Tree

```python
#importing DecisionTreeRegressor from sklearn library
from sklearn.tree import DecisionTreeRegressor
regressor=DecisionTreeRegressor(criterion='poisson',splitter='best',random_state=42)
regressor=regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
round(r_score,3)
```

#### 4. Random Forest

```python
#importing RandomForestRegressor from sklearn library
from sklearn.ensemble import RandomForestRegressor
regressor=RandomForestRegressor(n_estimators=100,criterion='squared_error',max_features='log2',random_state=42)
regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
round(r_score,3)
```

> The r2 score is better, **0.874**.
